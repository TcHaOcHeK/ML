```python
auto1 = pd.read_csv('auto1.csv')
auto2 = pd.read_csv('auto2.csv')
```

Создаем копию датафрейма auto1 не содержащую нулевых Unnamed столбцов

```python
auto1_copy = auto1.copy()
cut_coll = ['Unnamed: 9', 'Unnamed: 10', 'Unnamed: 11', 'Unnamed: 12', 'Unnamed: 13']
auto1_copy.drop(cut_coll, axis = 1, inplace = True)
auto1_copy.head()
```

Объединяем датафремы путем конкатинации.

```python
auto = pd.concat([auto1_copy, auto2], ignore_index=True)
print(auto.info())
```

Добавляем столбец 'owners' с рандомными значениями
в диапазоне от 10000 до 60000

```python
auto['owners'] = np.random.randint(10000, 60001, size=len(auto))
print(auto.head())
print(auto.info())
```

Определяем временной диапазон выпуска можелей датафрейма путем определения максимального и минимального значений столбца 'model'

```python
print(f'Ranged "model" {auto['model'].min()} - {auto['model'].max()}')
```

Определяем как менялся расход автомобилей в течении данного периода, группируя средние значения расзхода за каждый год периода

```python
auto.groupby('model')['mpg'].mean()
```

Выводим "Топ 5" предпочтительных моделей автомобилей за всё время

```python
auto['car'].value_counts().head(5)
```

Проверяем зависимость расхода от количнства целиндров группируя средние значеия расхода по количеству целиндров автомобилей.

```python
auto.groupby('cylinders')['mpg'].mean()
```

Докажем зависимость расхода автомобиля от веса, вычислив кореляцию значений.

результат "-0.832" показывает сильную обратную зависмость

```python
print(f"Корреляция: {auto['mpg'].corr(auto['weight']):.3f}")
print(f'\n {auto.groupby('weight')['mpg'].mean()}')
```

Сортируем датафрейм по значению столбца 'weight' по убыванию

```python
auto.sort_values(by = 'weight', ascending = False)
```
